```python
exchange_filtered = exchange[
    exchange["Series name"] == "Average Exchange rate of Pak Rupees per U.S. Dollar"
]
```

```python
exchange_clean = exchange_filtered[
    ["Observation Date", "Observation Value"]
].copy()
```

```python
exchange_clean["Date"] = pd.to_datetime(
    exchange_clean["Date"],
    format="%d-%b-%Y"
)
```

```python
exchange_clean = exchange_clean.sort_values("Date")
```
