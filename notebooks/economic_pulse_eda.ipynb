```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
master = pd.read_csv("../Data/Cleaned Data/master_dataset.csv")
```

```python
master["Date"] = pd.to_datetime(master["Date"])
```

```python
corr = master[
    [
        "Inflation_YoY",
        "USD_PKR",
        "Petrol_Price",
        "Remittances_Million_USD"
    ]
].corr()

plt.figure(figsize=(7,6))

plt.imshow(corr, cmap="coolwarm")

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=45
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(
            j,
            i,
            round(corr.iloc[i, j], 2),
            ha="center",
            va="center",
            color="black"
        )

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.savefig("../Figures/Correlation Heatmap.png", dpi=300)
plt.show()
```

```python
import matplotlib.dates as mdates

plt.figure(figsize=(12,5))

plt.plot(
    master["Date"],
    master["Inflation_YoY"]
)

plt.title("Inflation Trend (2010-2026)")
plt.xlabel("Year")
plt.ylabel("Inflation (%)")

# Show every year
ax = plt.gca()
ax.xaxis.set_major_locator(mdates.YearLocator(1))      # Every 1 year
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

plt.xticks(rotation=45)

plt.grid(True)
plt.tight_layout()

plt.savefig("../Figures/Inflation_Trend_Yearly.png", dpi=300)
plt.show()
```
