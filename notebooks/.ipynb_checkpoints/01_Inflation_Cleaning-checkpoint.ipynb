```python
cpi_part2_filtered = cpi_part2[
    cpi_part2["Series name"] ==
    "National CPI, an Inflation Measure (Year-on-Year basis)"
]
```

```python
cpi_part2_clean["Date"] = pd.to_datetime(
    cpi_part2_clean["Date"],
    format="%d-%b-%Y"
)
```
