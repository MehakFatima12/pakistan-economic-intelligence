```python
petrol_prices_2 = pd.read_csv(r"F:\Mehak Fatima\Pakistan-Economic-Intelligence-Platform\Data\Raw Data\Petrol Price.csv")
```

```python
petrol_prices_2 = petrol_prices_2.iloc[1:].copy()
```

```python
petrol_prices_2 = petrol_prices_2[
    ["Period End", "Pakistan"]
].copy()
```

```python
petrol_prices_2.head()
```

```python
petrol_prices_2.columns = [
    "Date",
    "Petrol_Price"
]
```

```python
petrol_prices_2["Date"] = pd.to_datetime(
    petrol_prices_2["Date"],
    format="%d-%b-%Y"
)
```

```python
petrol_prices_2["Month"] = petrol_prices_2["Date"].dt.to_period("M")
```

```python
petrol_prices_2.head()
```

```python
petrol_prices_2["Petrol_Price"] = pd.to_numeric(
    petrol_prices_2["Petrol_Price"],
    errors="coerce"
)
```

```python
petrol_prices_2["Petrol_Price"].dtype
```

```python
petrol_prices_2["Petrol_Price"].isnull().sum()
```

```python
monthly_petrol = (
    petrol_prices_2
    .groupby("Month")["Petrol_Price"]
    .mean()
)
```

```python
monthly_petrol.head
```

```python
monthly_petrol = monthly_petrol.reset_index()

monthly_petrol = monthly_petrol[
    monthly_petrol["Month"] >= "2025-06-01"
]
```

```python
petrol_prices.rename(
    columns={
        "Date": "Month",
        "Petrol Price (PKR per litre)": "Petrol_Price"
    },
    inplace=True
)
```

```python
petrol_final = pd.concat(
    [petrol_prices, monthly_petrol],
    ignore_index=True
)
```

```python
petrol_prices["Month"] = pd.to_datetime(petrol_prices["Month"])
```

```python
monthly_petrol["Month"] = monthly_petrol["Month"].dt.to_timestamp()
```

```python
petrol_final = pd.concat(
    [petrol_prices, monthly_petrol],
    ignore_index=True
)

petrol_final = petrol_final.sort_values("Month")
petrol_final.reset_index(drop=True, inplace=True)
```

```python
petrol_final.to_csv(
    r"F:\Mehak Fatima\Pakistan-Economic-Intelligence-Platform\Data\Cleaned Data\petrol_prices_clean.csv",
    index=False
)
```
